```python
import numpy as np

# Sample data as a numpy array
data = np.array([50, 750, 650, 750, 650, 750, 650, 750, 650, 750, 650, 750])
window_size = 3

# Create a kernel of ones, normalized by the window size (weights for the average)
weights = np.ones(window_size) / window_size

# Use numpy.convolve to apply the moving average kernel
# 'valid' mode returns only the fully computed results
sma_values_numpy = np.convolve(data, weights, mode='valid')

print("Original Data:", data)
print("SMA Values (NumPy Convolve):", sma_values_numpy.tolist())
# Output: [12.0, 14.0, 16.0, 18.0, 20.0, 22.0, 24.0, 26.0]
```
